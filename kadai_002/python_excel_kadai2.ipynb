```python

import pandas as pd

# データフレームの作成
df = pd.DataFrame({
    '日付':
      ['2023-05-17', '2023-05-18', '2023-05-19', '2023-05-20', '2023-05-21'],
    '社員名': ['山田', '佐藤', '鈴木', '田中', '高橋'],
    '売上': [100, 200, 150, 300, 250],
    '部門': ['メーカー', '代理店', 'メーカー', '商社', '代理店'],
})

# 売上の平均を求めて新しい列、「平均売上」列を作成
df['平均売上'] = df['売上'].mean()

# 売上で分岐して業績ランクを求める関数「performance」を定義
def performance(sales, average):
  result = ''
  if sales >= average + 50:
    result = 'Ａ'
  elif sales >= average:
    result = 'Ｂ'
  else:
    result = 'Ｃ'
  return result

# 「業績ランク」列を作成し、関数「performance」を適用して業績ランクを設定
df['業績ランク'] = df.apply(
    lambda row: performance(row['売上'], row['平均売上']),
    axis=1
)

# Excelファイルを作成
writer = pd.ExcelWriter('業績.xlsx')


# DataFrameオブジェクトをExcelファイルに書き込む
df.to_excel(writer, sheet_name='Sheet1', index=False)

# シートを取得  -----------  日付の一部か隠れるため、列幅調整
worksheet = writer.sheets['Sheet1']

# B列の日付列の幅を調整（例: 幅を15にする）
worksheet.column_dimensions['A'].width = 13

# Excelファイルを閉じる
writer.close()
```
