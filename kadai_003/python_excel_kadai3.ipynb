```python
import openpyxl
import pandas as pd
from openpyxl.styles import PatternFill
from openpyxl.utils.dataframe import dataframe_to_rows

# ---------------------------------------------  2022_年間売上表.xlsx、2023_年間売上表.xlsx　連結--------------
file_list = ['2022_年間売上表.xlsx', '2023_年間売上表.xlsx']

df = pd.DataFrame()

for file_name in file_list:

    temp_df = pd.read_excel(file_name)
    temp_df.columns = ['売上年', '商品', '金額（千円）']

    df = pd.concat([df, temp_df], ignore_index=False)

# ------------------------------------------------------------------------  グループ化           --------------
result = df.groupby(['商品', '売上年'])['金額（千円）'].sum().reset_index()

# -------------------------------------------------------------------------  売上集計表.xlsx　作成--------------
# 新しいワークブックを作成
wb = openpyxl.Workbook()
ws = wb.active

# グループ化したデータフレームからリストを取得してExcelファイルに追加
for r in dataframe_to_rows(result, index=False, header=True):
    ws.append(r)

# 塗りつぶしの設定
gray_fill = PatternFill(patternType='solid', fgColor='F2F2F2')
for col in range(1, ws.max_column + 1):
    ws.cell(row=1, column=col).fill = gray_fill

# 列幅を調整
ws.column_dimensions['A'].width = 15
ws.column_dimensions['C'].width = 13

wb.save('売上集計表.xlsx')
```
